## Import libraries


```python
import os
import numpy as np
import pandas as pd

CROP = "winter_wheat"
```

## Read the data


```python
phenology_df = pd.read_csv(f"phenology_{CROP}.csv")
phenology_df = phenology_df[phenology_df["STATE_NAME"].isin(["Berlin", "Brandenburg"])]
phenology_df.drop(columns=["PointID"], inplace=True)
phenology_df = phenology_df.groupby(by=["NUTS_ID", "harvest_year"]).mean(
    numeric_only=True
)
phenology_df.reset_index(inplace=True)
phenology_df[
    ["sowing_doy", "emergence_doy", "flowering_doy", "maturity_doy", "valid_sequence"]
] = phenology_df[
    ["sowing_doy", "emergence_doy", "flowering_doy", "maturity_doy", "valid_sequence"]
].astype(
    np.int64
)

print(phenology_df.shape)
phenology_df.head()
```

```python
# Check for the vIDPL
print(f"{CROP} vIDPL: {phenology_df['sowing_doy'].min()}")
```
